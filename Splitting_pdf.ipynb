```python
import fitz  # PyMuPDF

input_pdf  = "/content/Ivan Hubený_ Dimitri Mihalas - Theory of Stellar Atmospheres_ An Introduction to Astrophysical Non-equilibrium Quantitative Spectroscopic Analysis.pdf"
output_pdf = "/content/Cropped_Mihalas.pdf"

# Human-readable PDF page numbers: both ends are included.
start = 467
end   =540

source = fitz.open(input_pdf)

if not (1 <= start <= end <= len(source)):
    raise ValueError(
        f"Invalid range [{start}, {end}]. "
        f"This PDF has {len(source)} pages."
    )

clipped = fitz.open()

# PyMuPDF uses zero-based page indices internally.
clipped.insert_pdf(
    source,
    from_page=start - 1,
    to_page=end - 1
)

clipped.save(output_pdf)
clipped.close()
source.close()

print(f"Saved pages {start}–{end} to:\n{output_pdf}")
```
